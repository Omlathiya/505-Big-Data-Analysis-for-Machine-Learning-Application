## Step 0: Import libraries

```python
import string
from collections import Counter
```

## Step 1: Text corpus and stopword list

```python
corpus = [
    "Big data is the foundation of modern machine learning applications.",
    "The data pipeline cleans the data and prepares it for the model.",
    "Machine learning models learn patterns in the data and improve over time.",
    "Large language models are trained on big text data in the cloud.",
    "Data scientists use Python and Spark to process big data efficiently.",
]

STOPWORDS = {"the", "is", "in", "and", "of", "to", "for", "it", "on", "are", "a", "an",
             "over", "use", "with", "that", "this", "as", "at", "by"}
```

## Step 2: Cleaning function

```python
def clean(sentence: str) -> str:
    sentence = sentence.lower()
    for p in string.punctuation:
        sentence = sentence.replace(p, "")
    return " ".join(sentence.split())
```

## Step 3: Tokenize the corpus and filter stopwords

```python
all_tokens = []
for s in corpus:
    all_tokens.extend(clean(s).split())

filtered = [w for w in all_tokens if w not in STOPWORDS]

print(f"Total tokens (before filtering): {len(all_tokens)}")
print(f"Total tokens (after  filtering): {len(filtered)}")
print(f"Vocabulary size (unique words) : {len(set(filtered))}")
```

## Step 4: Word frequencies with `Counter`

```python
freq = Counter(filtered)
freq
```

## Step 5: Top 5 most frequent keywords

```python
for rank, (word, count) in enumerate(freq.most_common(5), 1):
    print(f"{rank}. {word:<10} {count}  {'#' * count}")
```
