## Step 0: Import libraries

```python
import string
```

## Step 1: Raw, noisy text samples

```python
raw_texts = [
    "   Hello!!! This is   a SAMPLE text, with NOISY punctuation...   ",
    "Big Data & Machine-Learning are TRANSFORMING industries (fast!).  ",
    "   Python's   simplicity: loved by Data Scientists; #DS505 @Surat   ",
]
for t in raw_texts:
    print(repr(t))
```

## Step 2: Cleaning function
1. `lower()` → lowercase  2. `replace()` → remove punctuation  3. whitespace removal + `strip()`

```python
def clean_text(text: str) -> str:
    text = text.lower()                       # 1. convert to lowercase
    for p in string.punctuation:              # 2. remove each punctuation symbol
        text = text.replace(p, "")
    text = " ".join(text.split())             # 3. collapse repeated whitespace
    return text.strip()                       #    strip leading / trailing spaces
```

## Step 3: Tokenization function – `split()`

```python
def tokenize(text: str) -> list:
    return text.split()
```

## Step 4: Apply cleaning and tokenization to every sentence

```python
for i, raw in enumerate(raw_texts, 1):
    cleaned = clean_text(raw)
    tokens = tokenize(cleaned)
    print(f"--- Sentence {i} ---")
    print("Raw     :", repr(raw))
    print("Cleaned :", repr(cleaned))
    print("Tokens  :", tokens)
    print("Count   :", len(tokens), "\n")
```
