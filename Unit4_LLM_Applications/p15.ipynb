## Step 1: Import libraries

```python
import pandas as pd
from transformers import pipeline
```

## Step 2: Load the pretrained sentiment-analysis pipeline

```python
classifier = pipeline("sentiment-analysis",
                      model="distilbert-base-uncased-finetuned-sst-2-english")
```

## Step 3: User review statements

```python
reviews = [
    "I absolutely love this phone, the battery life is fantastic!",
    "The delivery was late and the product was damaged. Very disappointed.",
    "Great value for money and excellent customer support.",
    "Worst purchase ever. It stopped working after two days.",
    "The movie was amazing, I would happily watch it again.",
    "The food was cold and the service was terrible.",
]
```

## Step 4: Classify the reviews

```python
results = classifier(reviews)
```

## Step 5: Display labels and confidence scores

```python
df = pd.DataFrame({
    "Review": reviews,
    "Sentiment": [r["label"] for r in results],
    "Confidence": [round(r["score"], 4) for r in results],
})
pd.set_option("display.max_colwidth", 80)
df
```

## Step 6: Summary

```python
pos = (df["Sentiment"] == "POSITIVE").sum()
print(f"{pos} POSITIVE, {len(df) - pos} NEGATIVE out of {len(df)} reviews")
```
