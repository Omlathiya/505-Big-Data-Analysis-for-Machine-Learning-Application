## Step 0: Import libraries

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql import functions as F
```

## Step 1: Start Spark and create the DataFrame

```python
spark = (SparkSession.builder.appName("DS505-Practical07").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Aarav",  "Department": "IT",      "Salary": 72000},
    {"Name": "Diya",   "Department": "HR",      "Salary": 54000},
    {"Name": "Kabir",  "Department": "Finance", "Salary": 65000},
    {"Name": "Meera",  "Department": "IT",      "Salary": 88000},
    {"Name": "Rohan",  "Department": "Sales",   "Salary": 48000},
    {"Name": "Isha",   "Department": "IT",      "Salary": 91000},
    {"Name": "Neha",   "Department": "HR",      "Salary": 58000},
    {"Name": "Vikram", "Department": "Finance", "Salary": 77000},
    {"Name": "Tara",   "Department": "Sales",   "Salary": 52000},
]
df = spark.createDataFrame([Row(**d) for d in data])
df.show()
```

## Step 2: Grouped aggregation + sorting (transformations are *lazy* – nothing runs yet)

```python
summary = (df.groupBy("Department")
             .agg(F.count("*").alias("Employee_Count"),
                  F.round(F.avg("Salary"), 2).alias("Avg_Salary"),
                  F.max("Salary").alias("Max_Salary"),
                  F.sum("Salary").alias("Total_Salary"))
             .orderBy(F.col("Total_Salary").desc()))
```

## Step 3: Execute actions – `.show()` and `.count()`
Actions trigger the actual computation.

```python
summary.show()

print("Total employees      (df.count())     :", df.count())
print("Number of departments (summary.count()):", summary.count())
```

## Step 4: Stop the session

```python
spark.stop()
```
