## Step 1: Import libraries

```python
from pyspark.sql import SparkSession, Row
```

## Step 2: Initialise the Spark session

```python
spark = (SparkSession.builder
         .appName("DS505-Practical05")
         .master("local[*]")                      # use all local CPU cores
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

print("Spark version:", spark.version)
```

## Step 3: Create a DataFrame from Python dictionaries

```python
data = [
    {"Name": "Aarav", "Department": "IT",      "City": "Surat",     "Salary": 72000, "Experience": 5},
    {"Name": "Diya",  "Department": "HR",      "City": "Mumbai",    "Salary": 54000, "Experience": 3},
    {"Name": "Kabir", "Department": "Finance", "City": "Surat",     "Salary": 65000, "Experience": 6},
    {"Name": "Meera", "Department": "IT",      "City": "Pune",      "Salary": 88000, "Experience": 9},
    {"Name": "Rohan", "Department": "Sales",   "City": "Ahmedabad", "Salary": 48000, "Experience": 2},
]
df = spark.createDataFrame([Row(**d) for d in data])
```

## Step 4: Schema tree hierarchy – `printSchema()`

```python
df.printSchema()
```

## Step 5: Print the rows – `show()`

```python
df.show()
```

## Step 6: Stop the session

```python
spark.stop()
```
