## Step 0: Import libraries

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col
```

## Step 1: Start Spark and create the DataFrame

```python
spark = (SparkSession.builder.appName("DS505-Practical06").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Aarav", "Department": "IT",      "City": "Surat",  "Salary": 72000},
    {"Name": "Diya",  "Department": "HR",      "City": "Mumbai", "Salary": 54000},
    {"Name": "Kabir", "Department": "Finance", "City": "Surat",  "Salary": 65000},
    {"Name": "Meera", "Department": "IT",      "City": "Pune",   "Salary": 88000},
    {"Name": "Rohan", "Department": "Sales",   "City": "Surat",  "Salary": 48000},
    {"Name": "Isha",  "Department": "IT",      "City": "Surat",  "Salary": 91000},
]
df = spark.createDataFrame([Row(**d) for d in data])
df.show()
```

## Step 2: Select specific columns – `.select()`

```python
df.select("Name", "City", "Salary").show()
```

## Step 3: Multi-condition filter – `.filter()`
`Salary > 60000` **and** `City == 'Surat'` (each condition must be in parentheses, joined with `&`).

```python
filtered = df.filter((col("Salary") > 60000) & (col("City") == "Surat"))
filtered.show()
```

## Step 4: Computed column – `.withColumn()`
Add `Annual_Bonus` = 10% of Salary.

```python
with_bonus = df.withColumn("Annual_Bonus", col("Salary") * 0.10)
with_bonus.show()
```

## Step 5: Rename a column – `.withColumnRenamed()`

```python
renamed = with_bonus.withColumnRenamed("Name", "Employee_Name")
renamed.show()
renamed.printSchema()
```

## Step 6: Stop the session

```python
spark.stop()
```
